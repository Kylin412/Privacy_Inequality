```python
# 调整后的 Firm Decision Factors 数据结构
# 1. 保持嵌套字典结构，便于按类别访问和扩展（保留层次信息）。
# 2. 新增一个扁平列表 'all_factors'，包含所有唯一的因素名称，按主类别 > 子类别 > 因素的顺序排序。
#    - 这个列表用于定义 one-hot 向量的维度和索引（每个因素对应一个位置）。
#    - 例如，all_factors[0] = "historical_performance"，对应 one-hot 向量的第 0 位。
# 3. 为什么这样调整：
#    - 嵌套字典：适合分类查询和维护层次（如分析内部 vs 外部因素）。
#    - 扁平列表：便于生成 one-hot 向量（向量长度 = len(all_factors)），每个公司的决策特征可以映射到这个向量。
#    - 方便共现分析：收集多个公司的 one-hot 向量后，可以用 numpy 计算共现矩阵（e.g., np.dot(oneshots.T, oneshots)）。

firm_decision_factors = {
    "Internal/Self-Oriented": {
        "Past/Historical": [
            "historical_performance",
            "price_stability_norm"
        ],
        "Future/Expected": [
            "profit_maximization",
            "demand_elasticity",
            "market_conditions_assessment"
        ]
    },
    "External/Surrounding-Oriented": {
        "Past/Historical": [
            "competitor_emulation"
        ],
        "Future/Expected": [
            "price_war_risk",
            "number_of_firms_competition",
            "predicted_share_rate",
            "share_rate_benefit"
        ]
    }
}

# 生成扁平的 all_factors 列表（所有唯一因素，按顺序扁平化）
all_factors = []
for main_category, sub_categories in firm_decision_factors.items():
    for sub_category, factors in sub_categories.items():
        all_factors.extend(factors)

all_factors
```
